# publish

```python
import asyncio
from aiomqtt import Client, MqttError

async def publish_temperature():
    # split host & port
    async with Client(
        hostname="artemis",  # from *inside* your app container
        port=1883,
        username="artemis",
        password="artemis",
    ) as client:
        await client.publish("temperature/outside", payload=b"28.4")
        # give the library a chance to flush
        await asyncio.sleep(1)

if __name__ == "__main__":
    try:
        await publish_temperature()
        print("✅ Published OK")
    except MqttError as e:
        print(f"❌ MQTT error: {e}")
```

# subscribe

```python
import asyncio
from aiomqtt import Client, MqttError

async def subscribe_temperature():
    async with Client(
        hostname="artemis",  # or "localhost" if running from your host
        port=1883,
        username="artemis",
        password="artemis",
    ) as client:
        # subscribe before messages are sent
        await client.subscribe("temperature/outside")
        print("Subscribed! Waiting for messages…")
        async for message in client.messages:
            print(f"[{message.topic}] {message.payload.decode()}")
            # break after one if you like:
            # break

if __name__ == "__main__":
    try:
        await subscribe_temperature()
    except MqttError as e:
        print(f"Error: {e}")
```
